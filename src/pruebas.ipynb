```python
from mlflow.tracking import MlflowClient
import mlflow
exchange = 'ETH-USD'
# Set the tracking URI to point to the MLflow server in Docker
tracking_uri = "http://localhost:5000"
mlflow.set_tracking_uri(tracking_uri)

# Initialize the MLflow client
client = MlflowClient(tracking_uri=tracking_uri)

# Define experiment name
experiment_name = f"ts_into_features_Daily_{exchange}"

# Get the experiment ID
experiment = client.get_experiment_by_name(experiment_name)
if experiment:
    experiment_id = experiment.experiment_id

    # Search for the runs within the experiment
    runs = client.search_runs(experiment_ids=[experiment_id])

    # Print out the metrics for each run
    for run in runs:
        print(f"Run ID: {run.info.run_id}")
        print(f"Metrics: {run.data.metrics}")
        print(f"Parameters: {run.data.params}")
else:
    print(f"Experiment '{experiment_name}' not found.")
```

```python
import streamlit as st
import requests
import pandas as pd
import plotly.graph_objects as go
import io
import graphviz


FASTAPI_URL = "http://127.0.0.1:8000"
CHARGE_MODEL = f"{FASTAPI_URL}/get_model"
REGISTER_MODEL = f"{FASTAPI_URL}/register_model"



# Function to fetch data from the FastAPI endpoint
def get_model(exchange):
    url = f"{CHARGE_MODEL}/{exchange}"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    else:
        st.error("There is no model registered")
        return None
    
#Function to register model
def register_model(exchange):
    url = f"{REGISTER_MODEL}/{exchange}"
    response = requests.post(url)  
    if response.status_code == 200:
        return response.json()
    else:
        st.error("Error registering the model")
        return None
```
